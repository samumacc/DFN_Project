## Download Datasets

```python
#!python download_datasets.py

import os
import zipfile
from google.colab import drive

# Monta il drive per accedere agli ZIP
if not os.path.exists('/content/drive'): drive.mount('/content/drive')

base_zip_path = '/content/drive/MyDrive/Datasets_VPR/'
local_path = '/content/datasets_vpr/' # Cartella locale veloce

datasets = {
    'Copia di gsv_xs.zip': 'gsv_xs',
    'Copia di sf_xs.zip': 'sf_xs',
    'Copia di tokyo_xs.zip': 'tokyo_xs',
    'Copia di robotcar_one_every_2m.zip': 'robotcar'
    }

if not os.path.exists(local_path):
  os.makedirs(local_path)

print("🚀 AVVIO ESTRAZIONE LOCALE (Veloce)...")
for zip_name, folder_name in datasets.items():
  zip_full_path = os.path.join(base_zip_path, zip_name)
  extract_full_path = os.path.join(local_path, folder_name)

  if os.path.exists(zip_full_path):
    print(f"⚡ Estraendo {zip_name}...")
    with zipfile.ZipFile(zip_full_path, 'r') as zip_ref:
      zip_ref.extractall(extract_full_path)
    print(f"✅ {folder_name} pronto in locale!")
  else:
    print(f"⚠️ File {zip_name} non trovato su Drive.")

print("\n✨ Fatto! Ora puoi iniziare a lavorare usando i percorsi in '/content/datasets_vpr/'")
```
